```python
# Create a new column 'Profitability' (Revenue per minute)
auto_data['Profitability'] = auto_data['Revenue (Millions)'] / auto_data['Runtime (Minutes)']

# Find the top 10 most profitable movies
top_10_profitable_movies = auto_data.sort_values(by='Profitability', ascending=False).head(10)

print("Top 10 Most Profitable Movies (Revenue per Minute):\n")
print(top_10_profitable_movies[['Movie_Title', 'Profitability']])
```

```python
# Define the conditions for categorization
conditions = [
    (auto_data['Rating'] < 6),
    (auto_data['Rating'] >= 6) & (auto_data['Rating'] < 7.5),
    (auto_data['Rating'] >= 7.5)
]

# Define the corresponding categories
choices = ['Poor', 'Average', 'Excellent']

# Create the new 'Rating_Category' column
auto_data['Rating_Category'] = np.select(conditions, choices, default='Unknown')

# Display the count of movies in each category
print("Distribution of movies by Rating Category:\n")
print(auto_data['Rating_Category'].value_counts())
```

## Custom Movie Scoring Function

I will now create a custom function to calculate a 'Movie_Score' for each movie based on its 'Rating' and 'Revenue (Millions)' using the formula: `Rating * 10 + Revenue (Millions)`. This function will then be applied to the DataFrame to create a new column.

```python
def calculate_movie_score(row):
    # Ensure 'Revenue (Millions)' is treated as 0 if NaN for scoring purposes
    revenue = row['Revenue (Millions)'] if pd.notna(row['Revenue (Millions)']) else 0
    return (row['Rating'] * 10) + revenue

# Apply the custom function to create the 'Movie_Score' column
auto_data['Movie_Score'] = auto_data.apply(calculate_movie_score, axis=1)

# Display the top 10 movies based on this new score
print("Top 10 Movies by Custom Score:\n")
print(auto_data[['Movie_Title', 'Rating', 'Revenue (Millions)', 'Movie_Score']].sort_values(by='Movie_Score', ascending=False).head(10))
```
