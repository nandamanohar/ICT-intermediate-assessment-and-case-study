```python
# 2. Weather Data Analyzer
# a. Analyze weather data (temperature, humidity, etc.) for a given region
# b. Use variables and data types to store weather data
# c. Implement conditional statements to determine weather conditions
# (sunny, rainy, etc.)
# d. Use functions to calculate average temperature and humidity, and
# print results



region = input("Enter the region: ")
temperature = [25, 32, 27, 37, 33]
humidity = [70, 89, 76, 68, 72]
weather = input("Enter weather condition (sunny/cloudy/rainy): ")

def avg_temperature():
    total = sum(temperature)
    average = total / len(temperature)
    return average

def avg_humidity():
    total = sum(humidity)
    average = total / len(humidity)
    return average

print("\nWeather Report")
print("Region:", region)
print("Weather:", weather)

if weather == "sunny":
    print("Sunny day.")
elif weather == "rainy":
    print("Rainy day.")
elif weather == "cloudy":
    print("Cloudy day.")
else:
    print("Weather not identified.")

print("Average Temperature:", avg_temperature())
print("Average Humidity:", avg_humidity())
```

```python
'''3. Password Validator
Write a function that checks if a password is strong: at least 8 characters,
contains at least one digit, one uppercase letter, and one special character
(!@#$%).'''



password = input("Enter a strong password ")

upper = False
lower = False
digit = False
special = False
spl_char = "!@#$%"
for letter in password:
  if letter.isupper():
    upper = True
  elif letter.islower():
    lower = True
  elif letter.isdigit():
    digit = True
  elif letter in spl_char:
    special = True
if len(password) >= 8 and upper and lower and digit and special:
  print("Password is STRONG!")
else:
  print("Password is WEAK!")
```

```python
# 4. Word Frequency Counter
# Given a paragraph of text, write a function that returns the 3 most
# frequently occurring words (ignore case and punctuation)

word = input("Enter a word: ")
letter_frequency = {}
for i in word:
    if i in letter_frequency:
        letter_frequency[i] += 1
    else:
        letter_frequency[i] = 1

print(letter_frequency)
# Sort by frequency
sorted_frequency = sorted(letter_frequency.items(), key=lambda x: x[1], reverse=True)
# Get top 3
top_3 = sorted_frequency[:3]
print(" most frequently occuring letter:", top_3)

```
